# Delta en Hull–White: Pathwise vs. Diferencias Finitas

**Objetivo del notebook.** Estimar la Delta de una opción europea bajo el modelo de Hull–White mediante dos métodos Monte Carlo:

1. **Diferencias Finitas centradas (FD)** con *common random numbers* (CRN).
2. **Pathwise (PW)** apoyado en la proporcionalidad $\partial S_T / \partial S_0 = S_T / S_0$ del esquema log-Euler.

Los dos estimadores son:

$$
\widehat{\Delta}^{\,FD} = \frac{\widehat{V}_0(S_0+h) - \widehat{V}_0(S_0-h)}{2h},
\qquad
\widehat{\Delta}^{\,PW} = \frac{1}{N}\sum_{n=1}^{N} D^{(n)}\, \mathbf{1}_{\{S_T^{(n)} > K\}}\, \frac{S_T^{(n)}}{S_0},
$$

donde $D^{(n)} = \exp\bigl(-\sum_{k=0}^{M-1} r_{t_k}^{(n)}\,\Delta t\bigr)$ es el factor de descuento estocástico aproximado.

El notebook contiene:

1. Estimación puntual con los parámetros de referencia del Cap. 4.
2. Comparación entre ambos métodos.
3. Sensibilidad del estimador FD respecto a $h$, frente a la independencia estructural del PW.

```python
import sys
sys.path.insert(0, "../src")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from hullwhite import GeneratePathsHWAndAssetEuler, P0T, PriceEuropeanOptionHWMC

np.set_printoptions(suppress=True, precision=6)
```

## 1. Parámetros del experimento

Se adoptan los parámetros estándar del Cap. 4 (Tabla de parámetros de referencia).

```python
S0        = 100.0
K         = 100.0
sigma_S   = 0.20
lambd     = 0.5
eta       = 0.01
rho       = -0.3
T         = 1.0
N         = 20_000
M         = 250
SEED      = 3
```

## 2. Estimadores

Se definen dos funciones encapsulando ambos métodos sobre la base de `GeneratePathsHWAndAssetEuler` y `PriceEuropeanOptionHWMC`.

```python
def DeltaFD_HW_MC(NoOfPaths, NoOfSteps, T, P0T, lambd, eta,
                  S0, sigma_S, rho, K, h=0.5,
                  option_type="call", seed=SEED):
    """
    Delta por diferencias finitas centradas en el modelo Hull-White
    con common random numbers (misma semilla en S0+h y S0-h).
    """
    paths_up = GeneratePathsHWAndAssetEuler(
        NoOfPaths, NoOfSteps, T, P0T, lambd, eta,
        S0 + h, sigma_S, rho, seed=seed)
    V_up, _ = PriceEuropeanOptionHWMC(paths_up, K, T, option_type)

    paths_dn = GeneratePathsHWAndAssetEuler(
        NoOfPaths, NoOfSteps, T, P0T, lambd, eta,
        S0 - h, sigma_S, rho, seed=seed)
    V_dn, _ = PriceEuropeanOptionHWMC(paths_dn, K, T, option_type)

    return (V_up - V_dn) / (2.0 * h)
```

```python
def DeltaPathwise_HW_MC(NoOfPaths, NoOfSteps, T, P0T, lambd, eta,
                       S0, sigma_S, rho, K,
                       option_type="call", seed=SEED):
    """
    Delta pathwise en Hull-White:
        Delta_PW = (1/N) sum_n D^(n) * 1_{S_T > K} * S_T / S0
    El factor D^(n) actúa como peso aleatorio constante respecto a S0.
    """
    paths = GeneratePathsHWAndAssetEuler(
        NoOfPaths, NoOfSteps, T, P0T, lambd, eta,
        S0, sigma_S, rho, seed=seed)

    R = paths["R"]
    S = paths["S"]
    dt = T / NoOfSteps

    # Descuento estocástico path-by-path
    integral_r = np.sum(R[:, :-1] * dt, axis=1)
    D = np.exp(-integral_r)
    ST = S[:, -1]

    if option_type.lower() == "call":
        indicator = (ST > K).astype(float)
    elif option_type.lower() == "put":
        indicator = -(ST < K).astype(float)
    else:
        raise ValueError("option_type debe ser 'call' o 'put'.")

    pathwise_realizations = D * indicator * (ST / S0)
    delta_pw = np.mean(pathwise_realizations)
    se_pw = np.std(pathwise_realizations, ddof=1) / np.sqrt(NoOfPaths)

    return delta_pw, se_pw
```

## 3. Estimación puntual

Se calculan los dos estimadores con los parámetros de referencia y un paso de perturbación $h = 0{,}5$ para FD.

```python
h = 0.5

delta_fd = DeltaFD_HW_MC(N, M, T, P0T, lambd, eta, S0, sigma_S, rho, K,
                         h=h, option_type="call", seed=SEED)

delta_pw, se_pw = DeltaPathwise_HW_MC(N, M, T, P0T, lambd, eta,
                                     S0, sigma_S, rho, K,
                                     option_type="call", seed=SEED)

print(f"Delta FD (h = {h})  = {delta_fd:.6f}")
print(f"Delta Pathwise     = {delta_pw:.6f}   (SE = {se_pw:.6f})")
print(f"|FD - PW|           = {abs(delta_fd - delta_pw):.6f}")
```

Los dos estimadores producen valores prácticamente coincidentes. La diferencia entre FD y PW es del orden del error estándar del estimador Pathwise, lo que confirma que ambos métodos están midiendo la misma cantidad. La extensión del esquema pathwise al modelo Hull–White se ha reducido a sustituir el descuento determinista $e^{-rT}$ por el descuento estocástico $D^{(n)}$ trayectoria a trayectoria, sin alterar la estructura del estimador desarrollada en el capítulo Black–Scholes.

## 4. Comparación de varianzas: FD vs. Pathwise

La coincidencia puntual entre ambos métodos no informa sobre su eficiencia relativa. Para compararla se repite la estimación de Delta sobre $n_{\text{rep}}$ simulaciones independientes (semillas distintas) manteniendo $N$ fijo. La desviación típica muestral de los $n_{\text{rep}}$ estimadores aproxima el error estándar de un estimador individual, y su cociente cuantifica la ganancia o pérdida de varianza de FD con CRN frente a Pathwise.


```python
# Repeticiones independientes para estimar la varianza de cada estimador
n_rep     = 60
base_seed = 1000   # semillas 1000, 1001, ... (distintas entre repeticiones)

def DeltaRepeticiones_HW(n_rep, base_seed, h=0.5):
    """
    Repite la estimacion de Delta sobre n_rep simulaciones independientes a N fijo.
    La desviacion tipica muestral sobre las repeticiones estima el error estandar
    de un estimador individual. Dentro de cada repeticion, FD emplea CRN (misma
    semilla en S0+h y S0-h); PW usa esa misma semilla.
    """
    fd = np.empty(n_rep)
    pw = np.empty(n_rep)
    for j in range(n_rep):
        s = base_seed + j
        fd[j] = DeltaFD_HW_MC(N, M, T, P0T, lambd, eta, S0, sigma_S, rho, K,
                              h=h, option_type="call", seed=s)
        pw[j], _ = DeltaPathwise_HW_MC(N, M, T, P0T, lambd, eta,
                                       S0, sigma_S, rho, K,
                                       option_type="call", seed=s)
    return fd, pw

fd_rep, pw_rep = DeltaRepeticiones_HW(n_rep, base_seed, h=0.5)

df_var = pd.DataFrame({
    "Metodo":    ["Diferencias finitas (CRN)", "Pathwise"],
    "Media":     [fd_rep.mean(), pw_rep.mean()],
    "SE (rep.)": [fd_rep.std(ddof=1), pw_rep.std(ddof=1)],
})

ratio = fd_rep.var(ddof=1) / pw_rep.var(ddof=1)
print(f"Repeticiones: {n_rep}   (N = {N}, M = {M})")
print(f"Cociente de varianzas FD/PW: {ratio:.4f}")
df_var.round(6)
```

Con $n_{\text{rep}}=60$ repeticiones, los dos estimadores presentan errores estándar prácticamente iguales (cociente de varianzas $\approx 1{,}00$). Para Delta, FD con CRN no mejora ni degrada la varianza respecto a Pathwise: el factor de descuento estocástico, idéntico en las dos simulaciones perturbadas, se cancela en la diferencia y deja a FD con una varianza equiparable a la del estimador directo. La ventaja de Pathwise es, por tanto, estructural (una sola simulación, sin elección de $h$) más que de eficiencia.


## 5. Sensibilidad del estimador FD respecto a $h$

Se realiza un barrido en el paso de perturbación $h$ para evaluar la estabilidad del estimador FD. El estimador PW se incluye como referencia horizontal.

```python
h_grid = np.array([10.0, 5.0, 2.0, 1.0, 0.5, 0.1, 0.05, 0.01, 0.005, 0.001])

results = []
for h in h_grid:
    d_fd = DeltaFD_HW_MC(N, M, T, P0T, lambd, eta, S0, sigma_S, rho, K,
                         h=h, option_type="call", seed=SEED)
    results.append({"h": h, "Delta_FD": d_fd, "|FD - PW|": abs(d_fd - delta_pw)})

df_h = pd.DataFrame(results)
df_h
```

```python
fig, ax = plt.subplots(figsize=(7, 4.5))

ax.semilogx(df_h["h"], df_h["Delta_FD"], marker="o", label=r"$\widehat{\Delta}^{FD}$")
ax.axhline(delta_pw, color="C1", linestyle="--", label=r"$\widehat{\Delta}^{PW}$")

ax.set_xlabel(r"$h$")
ax.set_ylabel(r"$\Delta$")
ax.set_title("Delta FD vs. Pathwise en Hull–White")
ax.legend()
ax.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

**Observaciones.**

1. **PW.** El estimador Pathwise no depende de $h$: aparece como una línea horizontal de referencia.
2. **FD.** Para $h$ moderado, el estimador FD coincide con PW a la precisión del error Monte Carlo. Para $h$ muy grande ($h = 10$), el sesgo de truncamiento $\mathcal{O}(h^2)$ se hace visible. Para $h$ pequeño el estimador permanece estable gracias al uso de CRN, que mantiene idénticas las trayectorias del tipo corto (y por tanto el descuento $D^{(n)}$) en las dos simulaciones perturbadas.
3. La estructura del comportamiento $h$-FD es idéntica a la observada en el capítulo Black–Scholes: el descuento estocástico no introduce nuevas patologías para esta griega.

## 6. Conclusiones

1. Para Delta de una call europea bajo Hull–White, **PW y FD con CRN producen estimadores de precisión comparable**, con una diferencia entre ambos del orden del error estándar Monte Carlo.
2. El estimador Pathwise mantiene su **ventaja estructural** respecto a FD: una sola simulación e independencia respecto a $h$.
3. El comportamiento cualitativo es idéntico al observado en Black–Scholes (Notebook 02). El factor de descuento estocástico $D^{(n)}$ actúa como peso aleatorio independiente de $S_0$ y no degrada la calidad de ninguno de los dos estimadores.
